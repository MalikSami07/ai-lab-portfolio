```python
import joblib

# Load the pipeline with zero manual preprocessing steps defined here
pipeline = joblib.load("sentiment_pipeline.pkl")

# A raw, unprocessed string -- straight from a hypothetical new article
sample_text = "This new AI tool is absolutely fantastic and makes content creation a breeze."

prediction = pipeline.predict([sample_text])[0]
probability = pipeline.predict_proba([sample_text])[0]

print("Raw input:", sample_text)
print("Predicted label:", prediction, "(1 = above median polarity, 0 = below)")
print("Prediction probabilities [below, above]:", probability)
```
