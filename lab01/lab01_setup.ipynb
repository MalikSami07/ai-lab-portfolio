```python
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv"
df = pd.read_csv(url)

print("Shape (rows, columns):", df.shape)
df.head()
```

```python
# 2. COMPOSITION -- column names, dtypes and non-null counts.
#    This is where missing values are discovered.
df.info()
```

```python
# 3. DISTRIBUTION -- count, mean, std, min, max and the 25/50/75 percentiles
#    for every numeric column. This reveals scale, spread and outliers.
df.describe()
```

```python
# Missing values, stated explicitly rather than inferred from info().
missing = df.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

print("Columns containing missing values:\n")
print(missing)
print()
print("As a percentage of all rows:\n")
print((missing / len(df) * 100).round(1))
```
